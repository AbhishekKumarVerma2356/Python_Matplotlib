## **Data Visualization**

#### **Data Visualization** ---> It helps us to extract **information and patterns from data using visuals and graphs**, which makes it easier to analyze important insights.

### **5 Types of Visuals**

#### **1. 2-D Plot** ---> Used to represent data using two dimensions.

#### **2. Scatter Plot** ---> Used to show the relationship between two numerical values.

#### **3. Bar Chart** ---> Used to compare values between different categories.

#### **4. Histogram** ---> Used to show the distribution of numerical data.

#### **5. Pie Chart** ---> Used to show the proportion or percentage of different categories.
