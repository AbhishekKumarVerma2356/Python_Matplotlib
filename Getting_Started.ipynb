### **Univariate Analysis & Bivariate Analysis**

#### **Univariate Analysis** ---> It is the analysis of **only one variable or one column at a time (Unique Column)** to understand its values, distribution, or patterns.

#### **Example** ---> Analyzing only the **Marks** column of 10 students.

#### **Bivariate Analysis** ---> It is the analysis of **two variables or two columns at a time** to understand the relationship between them.

#### **Example** ---> Analyzing the relationship between **Marks** and **Study Hours**.
#### Bivariate Analysis ---> 1. Numeric-Numeric 2. Numeric-Categorical 3. Categorical-Categorical

### **Multivariate Analysis**

#### **Multivariate Analysis** ---> It is the analysis of **three or more variables or columns at the same time** to understand the relationship and patterns between them.

#### **Example** ---> Analyzing **Study Hours, Marks, and Attendance** together to understand how they affect a student's performance.

### **plt.figure()** ---> It is used to **create a new figure or canvas** where we can plot our graph.
#### **Syntax** ---> `plt.figure()`
#### **Example** ---> `plt.figure(figsize=(10,5))`
----
### **plt.title()** ---> It is used to **add a title to the graph**.
#### parameter ---> color ---> what color the Title is going to have
#### **Syntax** ---> `plt.title("Title of Graph")`

#### **Example** ---> `plt.title("Student Marks")`
----
### **plt.ylabel()** ---> It is used to **add a label to the Y-axis** of the graph.
#### **Syntax** ---> `plt.ylabel("Y-axis Label")`

#### **Example** ---> `plt.ylabel("Marks")`
----
### **plt.xlabel()** ---> It is used to **add a label to the X-axis** of the graph.

#### **Syntax** ---> `plt.xlabel("X-axis Label")`

#### **Example** ---> `plt.xlabel("Student Name")`
----
### **plt.grid()** ---> It is used to **show grid lines** on the graph, which makes it easier to read the values.

#### **Syntax** ---> `plt.grid()`

#### **Example** ---> `plt.grid(True)`
----
### **plt.show()** ---> helps in represent the chart 

#### **Syntax** ---> `plt.show()`

#### **Example** ---> `plt.show()`
----
### **plt.xlim()** ---> It is used to **set the limits of the X-axis** of the graph.

#### **Syntax** ---> `plt.xlim(min_value, max_value)`

#### **Example** ---> `plt.xlim(0, 100)` ---> `plt.xlim(10, 1000)`


### **plt.ylim()** ---> It is used to **set the limits of the Y-axis** of the graph.

#### **Syntax** ---> `plt.ylim(min_value, max_value)`

#### **Example** ---> `plt.ylim(0, 500)`

### Customization For Line

#### Parameter:
#### 1. color ---> What is going to be color of the Line ---> for eg. red, green, blue OR HTML hash code

#### 2. marker ---> how point of intersection appears --> o , ^ , s , + , * , d , h , p

#### 3. linestyle ---> it will be defining the style of the line ---> '--' , '-., '-' , ':'

#### 4. linewidth ----> it shows the width of the line ---> 1,1.5,2,....

#### 5. markersize ---> it shows the markersize or it can change the size of the marker, default --> 6, ---> 10, 9,...

#### 6. label ---> label that what this specific line is representing but is only help while plotting multiple line in a single visualozation

### **groupby()**

#### **groupby()** ---> It helps to **group the rows based on the values of a particular column**. After creating groups, we can apply mathematical functions like `sum()`, `mean()`, `max()`, `min()`, etc. on the grouped data.

#### **Example**

#### Dataset ---> Country and Sales Information

| Country | Sales |
|---|---:|
| India | 150 |
| China | 5000 |
| USA | 1000 |
| India | 5000 |
| USA | 5000 |

#### **Country-wise Total Sales**

##### **Group 1** ---> `India` ---> `150 + 5000 = 5150`

##### **Group 2** ---> `China` ---> `5000`

##### **Group 3** ---> `USA` ---> `1000 + 5000 = 6000`

#### **Syntax** ---> `df.groupby("column_name")["value_column"].sum()`

### **Example** ---> `df.groupby("Country")["Sales"].sum()`
